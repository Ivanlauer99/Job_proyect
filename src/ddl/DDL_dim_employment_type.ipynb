```sql
-- Crear dimensión de tipos de empleo con clave surrogada autoincremental
CREATE OR REPLACE TABLE prueba_api.gold.dim_employment_type (
    employment_type_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    job_employment_type STRING,
    employment_type_es STRING,
    fecha_carga TIMESTAMP
)
COMMENT 'Dimensión de tipos de empleo con clave surrogada autoincremental (IDENTITY)';
```
