```sql
-- Crear dimensión de empleadores con clave surrogada autoincremental
CREATE OR REPLACE TABLE prueba_api.gold.dim_employer (
    employer_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    employer_name STRING,
    employer_logo STRING,
    employer_website STRING,
    employer_reviews STRING,
    fecha_carga TIMESTAMP
)
COMMENT 'Dimensión de empleadores con clave surrogada autoincremental (IDENTITY)';
```
