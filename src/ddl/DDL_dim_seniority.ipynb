```sql
-- Crear dimensión de seniority con clave surrogada autoincremental
CREATE OR REPLACE TABLE prueba_api.gold.dim_seniority (
    seniority_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    seniority_level STRING NOT NULL,
    seniority_level_es STRING,
    seniority_order INT,
    fecha_carga TIMESTAMP
)
COMMENT 'Dimensión de niveles de seniority/experiencia extraídos desde job_title';
```
