```sql
-- Crear dimensión de ubicaciones con clave surrogada autoincremental
CREATE OR REPLACE TABLE prueba_api.gold.dim_location (
    location_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    job_country STRING,
    job_state STRING,
    job_city STRING,
    location_full_text STRING,
    ubicacion_completa STRING,
    job_latitude DOUBLE,
    job_longitude DOUBLE,
    fecha_carga TIMESTAMP
)
COMMENT 'Dimensión de ubicaciones geográficas con clave surrogada autoincremental (IDENTITY)';
```
