```sql
-- ============================================
-- 🏠 VISTA DE KPIs POR MODALIDAD DE TRABAJO
-- ============================================
-- Compara métricas entre trabajo remoto vs presencial
-- Útil para tendencias de modalidad y análisis de flexibilidad

CREATE OR REPLACE VIEW prueba_api.semantic.vw_kpis_by_modality
COMMENT 'KPIs agregados por modalidad de trabajo (Remoto vs Presencial)'
AS
SELECT 
    -- Modalidad
    modalidad_trabajo,
    
    -- Métricas de volumen
    COUNT(*) AS total_ofertas,
    COUNT(DISTINCT employer_sk) AS empleadores_unicos,
    COUNT(DISTINCT location_sk) AS ubicaciones_unicas,
    COUNT(DISTINCT employment_type_sk) AS tipos_empleo_disponibles,
    
    -- Porcentaje del mercado
    ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS pct_del_mercado,
    
    -- Metadata
    MAX(fecha_carga) AS ultima_actualizacion
    
FROM prueba_api.semantic.vw_market_overview
GROUP BY modalidad_trabajo;
```

```sql
-- ============================================
-- EJEMPLO: COMPARATIVA REMOTO VS PRESENCIAL
-- ============================================
-- Compara métricas entre ambas modalidades

SELECT 
    total_ofertas
FROM prueba_api.gold.vw_kpis_by_modality
where modalidad_trabajo = 'Remoto'
```
