```sql
%sql
-- ============================================
-- 📊 VISTA: RANKING DE SKILLS
-- ============================================
-- Agrega el conteo de menciones por skill con métricas adicionales

CREATE OR REPLACE VIEW prueba_api.semantic.vw_skills_ranking
COMMENT 'Ranking de skills más demandadas con métricas salariales'
AS
WITH skill_totals AS (
  SELECT 
    -- Lenguajes
    SUM(has_python) AS python_count,
    SUM(has_sql) AS sql_count,
    SUM(has_scala) AS scala_count,
    SUM(has_java) AS java_count,
    SUM(has_r) AS r_count,
    
    -- Big Data
    SUM(has_spark) AS spark_count,
    SUM(has_hadoop) AS hadoop_count,
    SUM(has_kafka) AS kafka_count,
    SUM(has_flink) AS flink_count,
    
    -- Cloud
    SUM(has_aws) AS aws_count,
    SUM(has_azure) AS azure_count,
    SUM(has_gcp) AS gcp_count,
    SUM(has_databricks) AS databricks_count,
    SUM(has_snowflake) AS snowflake_count,
    
    -- ETL
    SUM(has_airflow) AS airflow_count,
    SUM(has_dbt) AS dbt_count,
    SUM(has_etl) AS etl_count,
    
    -- Data Warehouses
    SUM(has_redshift) AS redshift_count,
    SUM(has_bigquery) AS bigquery_count,
    
    -- Databases
    SUM(has_postgresql) AS postgresql_count,
    SUM(has_mysql) AS mysql_count,
    SUM(has_mongodb) AS mongodb_count,
    
    -- Containers
    SUM(has_docker) AS docker_count,
    SUM(has_kubernetes) AS kubernetes_count,
    
    -- AWS Services
    SUM(has_s3) AS s3_count,
    SUM(has_emr) AS emr_count,
    SUM(has_lambda) AS lambda_count,
    SUM(has_glue) AS glue_count,
    
    -- ML
    SUM(has_ml) AS ml_count,
    
    COUNT(*) AS total_ofertas
  FROM prueba_api.gold.vw_job_skills
)
SELECT 
  'Python' AS skill,
  'Lenguaje' AS categoria,
  python_count AS menciones,
  ROUND(python_count * 100.0 / total_ofertas, 2) AS pct_ofertas
FROM skill_totals
UNION ALL
SELECT 'SQL', 'Lenguaje', sql_count, ROUND(sql_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Scala', 'Lenguaje', scala_count, ROUND(scala_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Java', 'Lenguaje', java_count, ROUND(java_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'R', 'Lenguaje', r_count, ROUND(r_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Spark', 'Big Data', spark_count, ROUND(spark_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Hadoop', 'Big Data', hadoop_count, ROUND(hadoop_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Kafka', 'Big Data', kafka_count, ROUND(kafka_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Flink', 'Big Data', flink_count, ROUND(flink_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'AWS', 'Cloud', aws_count, ROUND(aws_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Azure', 'Cloud', azure_count, ROUND(azure_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'GCP', 'Cloud', gcp_count, ROUND(gcp_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Databricks', 'Cloud', databricks_count, ROUND(databricks_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Snowflake', 'Cloud', snowflake_count, ROUND(snowflake_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Airflow', 'ETL/Orchestration', airflow_count, ROUND(airflow_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'dbt', 'ETL/Orchestration', dbt_count, ROUND(dbt_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'ETL', 'ETL/Orchestration', etl_count, ROUND(etl_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Redshift', 'Data Warehouse', redshift_count, ROUND(redshift_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'BigQuery', 'Data Warehouse', bigquery_count, ROUND(bigquery_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'PostgreSQL', 'Database', postgresql_count, ROUND(postgresql_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'MySQL', 'Database', mysql_count, ROUND(mysql_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'MongoDB', 'Database', mongodb_count, ROUND(mongodb_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Docker', 'DevOps', docker_count, ROUND(docker_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Kubernetes', 'DevOps', kubernetes_count, ROUND(kubernetes_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'S3', 'AWS Services', s3_count, ROUND(s3_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'EMR', 'AWS Services', emr_count, ROUND(emr_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Lambda', 'AWS Services', lambda_count, ROUND(lambda_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Glue', 'AWS Services', glue_count, ROUND(glue_count * 100.0 / total_ofertas, 2) FROM skill_totals
UNION ALL
SELECT 'Machine Learning', 'ML/AI', ml_count, ROUND(ml_count * 100.0 / total_ofertas, 2) FROM skill_totals;
```

```sql
%sql
-- ============================================
-- EJEMPLO 1: TOP 20 SKILLS MÁS DEMANDADAS
-- ============================================
-- Ranking general de las skills más mencionadas en ofertas

SELECT 
    skill,
    categoria,
    menciones,
    pct_ofertas AS pct_del_mercado
FROM prueba_api.semantic.vw_skills_ranking
WHERE menciones > 0
ORDER BY menciones DESC
LIMIT 20;
```
