```sql
-- ============================================
-- 🏬 VISTA DE KPIs POR EMPLEADOR
-- ============================================
-- Agrega todas las métricas clave por empleador
-- Útil para rankings, comparaciones y dashboards de empleadores

CREATE OR REPLACE VIEW prueba_api.semantic.vw_kpis_by_employer
COMMENT 'KPIs agregados por empleador para análisis comparativo'
AS
SELECT 
    -- Identificadores
    employer_sk,
    employer_name,
    employer_website,
    
    -- Métricas de volumen
    COUNT(*) AS total_ofertas,
    COUNT(DISTINCT location_sk) AS ubicaciones_unicas,
    COUNT(DISTINCT employment_type_sk) AS tipos_empleo_ofrecidos,
    
    -- Métricas de modalidad
    SUM(CASE WHEN job_is_remote = true THEN 1 ELSE 0 END) AS ofertas_remotas,
    SUM(CASE WHEN job_is_remote = false THEN 1 ELSE 0 END) AS ofertas_presenciales,
    ROUND(AVG(CASE WHEN job_is_remote = true THEN 100.0 ELSE 0 END), 2) AS pct_remoto,
    
    -- Metadata
    MAX(fecha_carga) AS ultima_actualizacion
    
FROM prueba_api.gold.vw_market_overview
GROUP BY 
    employer_sk,
    employer_name,
    employer_website;
```

```sql
SELECT employer_name FROM prueba_api.gold.vw_kpis_by_employer
ORDER BY total_ofertas DESC
```

```sql
-- ============================================
-- EJEMPLO 1: TOP 10 EMPLEADORES
-- ============================================
-- Consulta simple usando la vista vw_kpis_by_employer

SELECT 
    employer_name AS empleador,
    employer_website AS sitio_web,
    total_ofertas,
    ubicaciones_unicas,
    tipos_empleo_ofrecidos,
    pct_remoto AS pct_trabajo_remoto,
    ofertas_remotas,
    ofertas_presenciales
FROM prueba_api.gold.vw_kpis_by_employer
ORDER BY total_ofertas DESC
LIMIT 10;
```
