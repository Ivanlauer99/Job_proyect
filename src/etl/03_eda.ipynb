```sql
-- Porcentaje de valores nulos por columna (convertir todos a STRING para el UNION)
WITH columnas_nulas AS (
    SELECT 'job_id' AS column_name, CAST(job_id AS STRING) AS value FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'employer_name', CAST(employer_name AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_title', CAST(job_title AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_description', CAST(job_description AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_country', CAST(job_country AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_salary', CAST(job_salary AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_is_remote', CAST(job_is_remote AS STRING) FROM prueba_api.bronze.jobs_bronze
    UNION ALL
    SELECT 'job_employment_type', CAST(job_employment_type AS STRING) FROM prueba_api.bronze.jobs_bronze
)
SELECT
    column_name,
    ROUND((SUM(CASE WHEN value IS NULL THEN 1 ELSE 0 END) * 100.0) / COUNT(*), 2) AS porcentaje_nulos
FROM columnas_nulas
GROUP BY column_name
ORDER BY porcentaje_nulos DESC;
```

```sql
SELECT
    COUNT(DISTINCT job_id) AS job_id_unicos,
    COUNT(DISTINCT employer_name) AS employer_name_unicos,
    COUNT(DISTINCT job_title) AS job_title_unicos,
    COUNT(DISTINCT job_description) AS job_description_unicos,
    COUNT(DISTINCT job_country) AS job_country_unicos,
    COUNT(DISTINCT job_salary) AS job_salary_unicos,
    COUNT(DISTINCT job_is_remote) AS job_is_remote_unicos,
    COUNT(DISTINCT job_employment_type) AS job_employment_type_unicos
FROM prueba_api.bronze.jobs_bronze;
```

```sql
SELECT
    COUNT(CASE WHEN LOWER(job_description) LIKE '%python%' THEN 1 END) AS trabajos_python,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%sql%' THEN 1 END) AS trabajos_sql,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%spark%' THEN 1 END) AS trabajos_spark,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%databricks%' THEN 1 END) AS trabajos_databricks,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%snowflake%' THEN 1 END) AS trabajos_snowflake,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%airflow%' THEN 1 END) AS trabajos_airflow,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%dbt%' THEN 1 END) AS trabajos_dbt,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%kafka%' THEN 1 END) AS trabajos_kafka,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%aws%' THEN 1 END) AS trabajos_aws,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%azure%' THEN 1 END) AS trabajos_azure,
    COUNT(CASE WHEN LOWER(job_description) LIKE '%gcp%' THEN 1 END) AS trabajos_gcp
FROM prueba_api.bronze.jobs_bronze;
```

```sql
CREATE OR REPLACE TEMP VIEW jobs_transformados AS
SELECT
    -- Identificadores
    job_id,
    
    -- Información del empleador
    employer_name,
    employer_logo,
    employer_website,
    employer_reviews,
    
    -- Información del trabajo
    job_title,
    job_description,
    job_publisher,
    job_employment_type,
    job_employment_types,
    
    -- Ubicación
    job_location,
    job_city,
    job_state,
    job_country,
    CONCAT(COALESCE(job_country, ''), ' - ', COALESCE(job_state, '')) AS ubicacion_completa,
    job_latitude,
    job_longitude,
    
    -- Trabajo remoto
    job_is_remote,
    
    -- Información salarial
    job_salary,
    job_salary_string,
    job_min_salary,
    job_max_salary,
    job_salary_period,
    
    -- Fechas
    job_posted_at,
    job_posted_at_timestamp,
    job_posted_at_datetime_utc,
    TRY_CAST(job_posted_at AS DATE) AS job_posted_at_date,
    
    -- Aplicación
    job_apply_link,
    job_apply_is_direct,
    apply_options,
    
    -- Beneficios
    job_benefits,
    job_benefits_strings,
    
    -- Clasificación ocupacional
    job_onet_soc,
    job_onet_job_zone,
    
    -- Enlaces
    job_google_link,
    
    -- Datos rescatados
    _rescued_data
    
FROM prueba_api.bronze.jobs_bronze;
-- Vista con TODAS las columnas de la tabla bronze
```
